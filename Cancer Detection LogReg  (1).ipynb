```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("data.csv")
```

```python
df.head()
```

```python
# The dataset appears to be related to breast cancer diagnosis, with columns representing various measurements and 
# a "diagnosis" column indicating whether the tumor is malignant ('M') or benign ('B').
df["diagnosis"].unique()
```

```python
df.describe()
```

```python
sns.heatmap(df.isnull())
```

```python
df.drop(["id","Unnamed: 32"], axis=1, inplace=True)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for i in df.select_dtypes(include='object').columns:
    df[i]=le.fit_transform(df[i])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
selected_features = ['radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst','smoothness_worst', 'compactness_worst', 'concavity_worst','concave points_worst', 'symmetry_worst', 'fractal_dimension_worst']
x = df[selected_features]
y=df["diagnosis"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
reg=LogisticRegression()
```

```python
reg.fit(x_train,y_train)
```

```python
y_pred=reg.predict(x_test)
```

```python
x_test
```

```python
y_test
```

```python
reg.predict([[14.39, 17.70, 105.00, 639.1, 0.1254, 0.58490, 0.7727, 0.15610, 0.2639, 0.11780]]) # perfect prediction
```

```python
# Evaluate the model
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f'Accuracy: {accuracy*100}','%')
```

```python
print('Confusion Matrix:')
print(conf_matrix)
```

```python
print('Classification Report:')
print(class_report)
```

```python
plt.plot(y_test,y_test,color='black')
plt.plot(y_pred,y_pred,linestyle='dashed', color='yellow')
plt.title('Performance of the model')
plt.xlabel('Actual Outome')
plt.ylabel('Predicted Outome')
plt.grid(True)
plt.show()
```
